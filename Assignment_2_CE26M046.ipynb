Q1.(vi) Plot global error for both the Explicit and Implicit method

```python
import numpy as npy
import matplotlib.pyplot as plt

# Make plots appear in Colab
%matplotlib inline

# Parameters
c = 0.5
m = 1
zeta = c / m
v0 = 1

# Time step sizes to test (logarithmically spaced)
dts = [1 * 0.1**(n-1) for n in range(1, 6)] # TODO print and see the values in "dts" array
print("Time step sizes (dts):", dts)

# Storage for errors
errors_explicit = []
errors_implicit = []

# Loop over dt values
for dt in dts:
    time = np.arange(0, 5 + dt, dt)   # time array

    # Exact solution
    velocity_exact = v0 * np.exp(-zeta * time)

    # Explicit Euler
    velocity_explicit = v0 * (1 - zeta*dt) ** np.arange(len(time))

    # Implicit Euler
    velocity_implicit = v0 * (1 / (1 + zeta*dt)) ** np.arange(len(time))

    # RMS errors --> TODO what is RMS? While calculating errors, we may get negative and positive errors. If we want to show what is the average error it gives zero because the negative and positive errors cancel each other out. To avoid this, we can take the square of the errors, then take the mean, and finally take the square root to get the RMS error.
    err_exp = np.sqrt(np.mean((velocity_explicit - velocity_exact)**2))
    err_imp = np.sqrt(np.mean((velocity_implicit - velocity_exact)**2))  # TODO what should be the expression here?

    errors_explicit.append(err_exp)# TODO what is this command doing? check! It takes the current value of error_exp and add it to the list of errors_explicit, which will be used later for plotting the error vs time step size.
    errors_implicit.append(err_imp)

# --- Log-Log Plot ---
plt.figure(figsize=(8,6))
plt.loglog(dts, errors_explicit, 'o-', label="Explicit Euler")
plt.loglog(dts, errors_implicit, 's-', label="Implicit Euler")

plt.xlabel("Time step size (dt)", fontsize=14)
plt.ylabel("Global RMS Error", fontsize=14)
plt.title("Error vs Time Step (Log-Log)", fontsize=16)
plt.legend(fontsize=12)
plt.grid(True, which="both", ls="--")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# inputs
length = 1
area = 0.01
E = 2e11
b0 = -1000 / (E * area)  # axial load term
u0 = 0
uL = 0

dx2 = np.array([0.2, 0.1, 0.05, 0.025])        # for error plot, TODO choose mesh sizes for log plot

# ---------- Figure 2: RMSE vs dx (using dx2) ----------
rms_error = np.zeros_like(dx2, dtype=float)

for i, h in enumerate(dx2): 
    print(i, h) # TODO check what "enumerate" does
    n = int(round(length / h)) + 1
    x = np.linspace(0, length, n)
    print(x)
    h_eff = length / (n - 1)

    main = -2 * np.ones(n - 2)
    up   =  1 * np.ones(n - 3) # TODO check the size of this array and compare with size of "main"
    low  =  1 * np.ones(n - 3)
    print("Size of main:", main.size, "Size of up:", up.size, "Size of low:", low.size) # TODO check the sizes of these arrays
    A = np.diag(main) + np.diag(up, 1) + np.diag(low, -1) # TODO what is the size of this matrix for each mesh size
    print("Size of A:", A.shape)
    b = b0 * h_eff**2 * np.ones(n - 2)
    b[0] -= u0; b[-1] -= uL

    u_int = np.linalg.solve(A, b)
    u_nodes = np.concatenate(([u0], u_int, [uL]))

    u_exact_nodes = -b0/2 * x * (1 - x)
    err = u_nodes - u_exact_nodes
    rms_error[i] = np.sqrt(np.mean(err**2)) # TODO can you re-try this with average absolute error or maximum absolute error instead of RMS?

plt.figure()
plt.loglog(dx2, rms_error, 'o--')
plt.xlabel("nominal dx"); plt.ylabel("RMSE")
plt.title("RMSE vs dx (log-log plot)")
plt.grid(True, which='both')
plt.show()
```
